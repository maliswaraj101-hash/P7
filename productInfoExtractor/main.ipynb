```python
TRAIN_DATA = [
  ("Samsung Galaxy M14 5G with 6000mAh battery and 128GB storage", {
    "entities": [(0, 7, "BRAND"), (8, 21, "MODEL"), (27, 35, "BATTERY"), (44, 50, "STORAGE")]
  }),
  ("LG 190 L 4 Star Inverter Single Door Refrigerator, Blue Plumeria", {
    "entities": [(0, 2, "BRAND"), (3, 9, "CAPACITY"), (10, 17, "ENERGY_RATING"), (44, 64, "TYPE"), (66, 80, "COLOR")]
  }),
  ("Philips 9W LED Bulb, Cool Daylight, Pack of 3", {
    "entities": [(0, 7, "BRAND"), (8, 11, "WATTAGE"), (12, 21, "TYPE"), (23, 36, "COLOR")]
  }),
  ("Bajaj 1000W Mixer Grinder with 3 jars and stainless steel blades", {
    "entities": [(0, 5, "BRAND"), (6, 11, "WATTAGE"), (12, 25, "PRODUCT_TYPE"), (50, 68, "MATERIAL")]
  }),
]
```

```python
import spacy
```

```python
nlp = spacy.blank("en")
# blank spacy pipeline without tokenier  , NER , it is not pretrained
ner = nlp.add_pipe("ner")
# adds NER component to pipeline
```

```python
for _,annotation in TRAIN_DATA:
    print(annotation)
```

```python
for _,annotation in TRAIN_DATA:
    print(annotation.get("entities"))
```

```python
for _, annotation in TRAIN_DATA:
    for ent in annotation.get("entities"):
        ner.add_label(ent[2])

# ent[2] is the entity label (like BRAND, MODEL, etc.)

# ner.add_label() tells spaCy what labels to expect — must be added before training starts.

```

```python
nlp.begin_training()
# Initializes the training process (optimizer setup, weight initialization).

```

```python
from spacy.training.example import Example
```

```python
for i in range(30):
    # epochs
    losses = {}
    for text , annotation in TRAIN_DATA:
        doc = nlp.make_doc(text)
        #  Converts raw text into a Doc object (tokenized), but without predicting entities yet.
        example = Example.from_dict(doc , annotation)
        # Associates the input doc (doc) with the expected output (entities in annotations).
        nlp.update([example], losses=losses)
        # Performs one training step (gradient update) on the NER model using the annotated example.
        # The loss for this batch is recorded into the losses dictionary.
    print(f"Losses at iteration {i}: {losses}")





        
```

```python
nlp.to_disk("ecommerce_ner_model")
```

```python
nlp = spacy.load("ecommerce_ner_model")
text = "Samsung Galaxy M14 5G with 6000mAh battery and 128GB storage"
doc = nlp(text)
```

```python
for ent in doc.ents:
    print(ent.text , "->" , ent.label_)
```
