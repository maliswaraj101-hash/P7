```python
import numpy as np
```

```python
matrix = [[1,2,3],
[4,5,6],
[7,8,9]]
```

```python
# say u want to subtract 3 from first row 6 from second and 9 from third
```

```python
arr = np.array([3,6,9])
```

```python
arr.shape
```

```python
matrix - arr
```

```python
# what happened here is  [[1,2,3],
# [4,5,6],
# [7,8,9]] ----- [3,6,9]
#                [3,6,9]
#                [3,6,9]

# python pads (3,) to (1,3) while broadcasting and thought of it as a row vector , and broadcasted it likewise
```

```python
# therefore we need to keep the dimensions if we want to keep the broadcasting proper
```

```python
matrix - np.max(matrix , axis=1)
```

```python
arr = np.reshape(arr , (3,1))
```

```python
matrix - arr
```

```python
matrix - np.max(matrix , axis=1 , keepdims=True)
```
