DENSE LAYER IMPLEMENTATION

```python
class Layer_Dense:
    def __init__(self , inputs , neurons):
        self.weights = 0.01 * np.random.randn(neurons , inputs)
        self.biases = np.zeros((1 , neurons))
    
    def forward_pass(self , inputs):
        self.output = np.dot(inputs , np.transpose(self.weights)) + self.biases

        
        
```

in problems like multi class classification , no of output neurons  = no of classes , each neuron gives a number , lets say we have 3 classes which leads to output layer having 3 neurons , each output from neuron is [5 , 10 , 15] , we need to find looking at this output what does our feature which was input in the neural netowrks , what class does it belong to , we will apply , softmax fucntion , which squishes value between (0,1) , what it does is for lets say 5 == e^5/(e^5 + e^10 + e^15) , same for 10 and 15 , converts them to probabilites , the one with the highest probability is the class that the feature belong to , the raw scores [5 , 10 , 15] are called logits.

```python
output = [5 , 10 ,15]

def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / np.sum(e_x)
```

```python
logits = np.array(output)
probs = softmax(logits)
```

```python
probs
```

```python
# x = np.array([5, 10, 15])
# np.max(x) = 15
# x - np.max(x) = [-10, -5, 0]
# np.exp([-10, -5, 0]) = [4.54e-05, 6.74e-03, 1.0]

# So e_x becomes [0.0000454, 0.00674, 1.0].

# np.sum(e_x) adds up all the exponentials.
# sum = 0.0000454 + 0.00674 + 1.0 ≈ 1.00678
# softmax = [0.0000454/1.00678, 0.00674/1.00678, 1.0/1.00678] ≈ [0.000045, 0.0067, 0.9933]
```

sigmoid is used for binary classifiation , where output neuron is just one , converts the logit to a value between 0 and 1 , and we can use a threshold  , lets say 0 if x <= 0.5 else 1 , therefore the value which sigmoid producs is compared with 0.5 and then its either 0 or 1

```python
class Relu:
    def applyRelu(self , inputs):
        self.output = np.maximum(0 , inputs)
```

```python
class Softmax:
    def applySoftmax(self , inputs):
        e_inputs = np.exp(inputs - np.max(inputs , axis=1 , keepdims=True))
        self.output = e_inputs / np.sum(e_inputs , axis=1 , keepdims=True) 
```

```python

class Loss_CategoricalCrossEntropy:
    def calculate(self, y_pred, y_true):
        # Clip data to prevent division by 0 or log(0)
        y_pred_clipped = np.clip(y_pred, 1e-15, 1 - 1e-15)

        # If labels are one-hot encoded
        if len(y_true.shape) == 2:
            # Use sum of only correct class probabilities
            correct_confidences = np.sum(y_pred_clipped * y_true, axis=1)
        else:
            # If labels are label-encoded (integers)
            correct_confidences = y_pred_clipped[range(len(y_pred)), y_true]

        #   negative log likelihoods
        negative_log_likelihoods = -np.log(correct_confidences)

        # Return mean loss across all samples
        return np.mean(negative_log_likelihoods)
```

```python
loss = Loss_CategoricalCrossEntropy()
y_pred = np.array([
    [0.7, 0.2, 0.1],
    [0.1, 0.1, 0.8],
    [0.2, 0.6, 0.2]
])
y_true = np.array([
    [1, 0, 0],
    [0, 0, 1],
    [0, 1, 0]
])
```

```python
loss.calculate(y_pred , y_true)
```

ACCURACY

```python
def accuracy(y_true , y_pred):
    y_pred = np.argmax(y_pred , axis=1)
    if(len(y_true.shape) == 2):
        y_true = np.argmax(y_true , axis=1)

    return np.mean(y_pred == y_true)    
```
