```python
import numpy as np
```

```python
weights = np.array([-3 , -1 , 2])
bias = 1
inputs = np.array([1 , -1 , 3])
target_output = 0
learning_rate = 0.001
```

```python
def relu(x):
    return np.maximum(0 , x)

def relu_derivative(x):
    return np.where(x > 0 , 1 , 0)

```

```python
iterations = 200


linear_output = np.dot(inputs , weights) + bias
relu_output = relu(linear_output)
loss = np.square(relu_output - target_output)
```

```python
# we need to find dL/dweights and dL/dbias , how loss is sensitive to each of them
```

```python
dL_drelu = 2 * (relu_output - target_output)
```

```python

drelu_dz = relu_derivative(linear_output)
```

```python
dz_dw = inputs
```

```python
dz_dbias = 1
```

```python
dL_dW = dL_drelu * drelu_dz *dz_dw
dL_dbias = dL_drelu * drelu_dz *dz_dbias
```

```python
weights = weights - learning_rate*dL_dW
bias = bias - learning_rate*dL_dbias
```

```python
#writing it in one loop
```

```python
iterations = 200

for i in range(iterations):
    linear_output = np.dot(inputs , weights) + bias
    relu_output = relu(linear_output)
    loss = np.square(relu_output - target_output)

    dL_drelu = 2 * (relu_output - target_output)
    drelu_dz = relu_derivative(linear_output)
    dz_dw = inputs
    dz_dbias = 1


    dL_dW = dL_drelu * drelu_dz *dz_dw
    dL_dbias = dL_drelu * drelu_dz *dz_dbias

    weights = weights - learning_rate*dL_dW
    bias = bias - learning_rate*dL_dbias

    print(f'iteration - {i + 1} loss is {loss}')

print(f'final weights {weights}')
```
