```python
from langchain_openai import OpenAIEmbeddings
from langchain.vectorstores import Chroma
from langchain.schema import Document
```

```python
# In chroma there is a tenent(user) , that user can create multiple databases
# each database can have multiple tables (Collection) , and in each collection we can store docs , each doc contains a embedding and its respective metdata.
```

```python
doc1 = Document(
    page_content="this is Virat Kohli",
    metadata = {"team" : "RCB"}
)

doc2 = Document(
    page_content="this is Rohit Sharma",
    metadata = {"team" : "MI"}
)

doc3 = Document(
    page_content="this is MS Dhoni",
    metadata = {"team" : "CSK"}
)
```

```python
docs = [doc1 , doc2 , doc3]
```

```python
vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory='chroma_db',
    # where the vectors will be stored
    collection_name='sample',
)


# Database = chroma_db (folder)
# Collection = sample
```

```python
vector_store.add_documents(docs)
# docs will be added to the "sample" collection inside the chroma_db database folder.
```

```python
vector_store.get(include=['embeddings' , 'documents' , 'metdata'])
# goes into the DB --> collection and then fetches the above mentioned info from the collection ( DB + COLLECTION ) pair is fixed for a chroma instance vector_store
```

```python
# one Chroma instance , that is chroma() , it creates a vector store object , which points to the specified (database + collection name) , this vector store object only deals with that particular DB and collection , if we want a new DB or new collection in new db or new collection in same db , we need to make another chroma instance vector store object , that will point to the (DB + collection) pair
```

```python
# One Chroma() instance = one vector store object.

# That vector store object is bound to one database folder (persist_directory) + one collection (collection_name).

# That object can only read/write/search within that one collection in that one database.

# If you need:

# A new database → use a new persist_directory and create a new Chroma() instance.

# A new collection in the same database → use the same persist_directory but a different collection_name, with a new Chroma() instance.

# So yes → every (DB + collection) pair you want to work with requires its own Chroma() object.
```

```python
vector_store.similarity_search(
    query='who among these is a bowler',
    k = 2
)
# k = 2 , will give u 2 similar docs , k = 1 , will give only 1 similar doc
```

```python
vector_store.similarity_search_with_score(
     query='who among these is a bowler',
    k = 2
)
#  will show similairty socre ( lesser the better , shows the distance , less distance means vectors are close means more similar.)
```

```python
vector_store.similarity_search_with_score(
     query='',
     filter={'team' : 'CSK'}
)
#  we can even filter based on the metadata
```

```python
# update documents

updated_doc = Document(
    page_content='this is a new doc',
    metadata = {'team' : "IDK"}
)

vector_store.update_document(document_id='provide the id of the document which we want to update' , document = updated_doc)

```

```python
# delete documents

vector_store.delete(ids= ['give the id of a document' , 'can give one more id of another document'])
```
