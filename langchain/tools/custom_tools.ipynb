```python
from langchain_core.tools import tool
```

```python
# step 1 -> create a function
def multiply(a , b):
    """Multiply 2 numbers"""
    return a * b
```

```python
# step 2 -> give types to it
def multiply(a : int , b : int) -> int:
    """Multiply 2 numbers"""
    return a * b
```

```python
# step 2 -> add tool decorator

@tool
def multiply(a :int , b : int) -> int:
    """Multiply 2 numbers"""
    return a * b
```

```python
result = multiply.invoke({"a" : 5 , "b" : 3})
```

```python
print(result)
```

```python
print(multiply.name)
print(multiply.description)
print(multiply.args)
# every tool has these attributes
# 
```
