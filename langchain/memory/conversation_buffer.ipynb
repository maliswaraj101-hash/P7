```python
from langchain_huggingface import ChatHuggingFace , HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.schema.runnable import RunnableSequence
from langchain.tools import BaseTool 
from typing import Type
from pydantic import BaseModel , Field


load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id = "google/gemma-2-2b-it",
    # repo_id="deepseek-ai/DeepSeek-R1-0528",

    # huggingfacehub_api_token="YOUR_HF_TOKEN",
    huggingfacehub_api_token="YOUR_HF_TOKEN",
    task = "text-generation"
)

model = ChatHuggingFace(llm = llm)
```

```python
from langchain.memory import ConversationBufferMemory , ConversationSummaryMemory , ConversationBufferWindowMemory , ConversationSummaryBufferMemory
from langchain.chains import ConversationChain
```

```python
memory = ConversationBufferMemory()
# memory = ConversationSummaryMemory()
# conversation summary memory does the same work as conversation buffer memory but the only difference is that conversation summary memory sends the summary of past convos not the whole convo like conversationbuffermemory
```

```python
window_memory = ConversationBufferWindowMemory(k=2)

# only past 2 Q/A will be stored and given to LLM with the query of the user

ConversationSummaryBufferMemory(max_token_limit=200)

# what this does is it keeps the past responses to 200 tokens and everything before that is summarised and fed with the user new query
# therefore RAW PAST 200 token + summary of data before 200 tokens + user query is sent to the LLM

conversation = ConversationChain(
    llm = llm,
    verbose = True,
    memory = memory,
    # memory = window_memory
    # use window_memory , if u want just past "k" Q/A pairs
)
# just logs the conversation 
# 
```

```python
conversation.predict(input = "hello there deepseek")
# we will get a response for this and when we again do conversation.predict() , conversationBufferMemory gives the LLM old Q/A + the current query , it keeps storing Q/A


#DISADVANTAGE
# keeps providing the whole convo , there will be a point where
# we will exceed the CONTEXT WINDOW of the LLM 
```
