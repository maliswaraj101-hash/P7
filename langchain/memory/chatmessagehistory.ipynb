```python
from langchain_huggingface import ChatHuggingFace , HuggingFaceEndpoint
from dotenv import load_dotenv
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain.schema.runnable import RunnableSequence
from langchain.tools import BaseTool 
from typing import Type
from pydantic import BaseModel , Field
from langchain.memory import ChatMessageHistory 


load_dotenv()

llm = HuggingFaceEndpoint(
    repo_id = "google/gemma-2-2b-it",
    # repo_id="deepseek-ai/DeepSeek-R1-0528",

    # huggingfacehub_api_token="YOUR_HF_TOKEN",
    huggingfacehub_api_token="YOUR_HF_TOKEN",
    task = "text-generation"
)

model = ChatHuggingFace(llm = llm)
```

```python
prompt = PromptTemplate(
    template="Tell me about yourself , this is the conversation history - {message_history}",
    input_variables=['message_history']
)
```

```python
parser = StrOutputParser()
```

```python
chain = prompt | model | parser
```

```python
message_history = ChatMessageHistory()
# message_history = InMemoryChatMessageHistory()
# both are the same
```

```python
message_history.add_user_message("hello my name is swaraj")
message_history.add_ai_message("Hi I am google")
```

```python
chain.invoke({f"{message_history.messages}"})
```
